```python
%%writefile valvar.l

%{
#include "y.tab.h"
%}

%%

[a-zA-Z]    { return LET; }
[0-9]       { return DIG; }

\n          { return 0; }

.           { return yytext[0]; }

%%

int yywrap()
{
    return 1;
}
```

```python
%%writefile valvar.y

%{
#include <stdio.h>
#include <stdlib.h>

int yylex();
int yyerror(const char *);
%}

%token LET DIG

%%

variable: var
         ;

var: var DIG
   | var LET
   | LET
   ;

%%

int main()
{
    printf("Enter the variable:\n");
    yyparse();
    printf("Valid variable\n");
    return 0;
}

int yyerror(const char *s)
{
    printf("Invalid variable\n");
    exit(0);
}
```
